Find the intersection of two lists (common elements)

```python
def find_intersection(list_1, list_2):
    final_list = []

    for x in list_1:
        if x in list_2:
            final_list.append(x)
        else:
            continue
    return final_list

find_intersection([1,2,3,4],[2,4,6,8])
```

```python
# No need of continue since loop continues without it
# Turning one list to a set is faster

def find_intersection(list_1, list_2):
    set_2 = set(list_2)
    result = []

    for x in list_1:
        if x in set_2:
            result.append(x)
        
    return result

find_intersection([1,2,3,4],[2,4,6,8])
```

```python
def find_intersection(list_1, list_2):
    set_2 = set(list_2)
        
    return [x for x in list_1 if x in set_2]

find_intersection([1,2,3,4],[2,4,6,8])
```

```python
# Using set methods - Order is lost and duplicates are removed

def find_intersection(list_1, list_2):

    set_1 = set(list_1)
    set_2 = set(list_2)

    return set_1.intersection(set_2)

find_intersection([1,2,3,4],[2,4,6,8])
```

Remove all occurrences of a specific value from a list

```python
# If value is removed from the input_list while iterating elements are skipped. 
# Should not use remove while iterating the same
```

```python
def remove_value(input_list, value):

    output_list = []
    for x in input_list:
        if x != value:
            output_list.append(x)
    return output_list

remove_value(["apple","banana"], "apple")
```

```python
remove_value([1, 2, 2, 2, 3], 2)
```

```python
def remove_value(input_list, value):
    return [x for x in input_list if x != value]
```

Find the missing number in a list containing numbers from 1 to n

```python
def find_missing_number(nums, n):
    return list((set(range(1, n + 1)) - set(nums)))

find_missing_number([1,2,4,6,7,9],10)
```

 Move all zeros in a list to the end while maintaining order of other elements

```python
def arrange_zeros(arr):
    result = []
    zero_list = []

    for num in arr:
        if num == 0:
            zero_list.append(0)
        else:
            result.append(num)
    result.extend(zero_list)

    return result

arrange_zeros([1,2,0,0,4,0,5,6])
```

```python
def arrange_zeros(arr):
    result = [num for num in arr if num != 0]
    zero_list = [0]*(len(arr) - len(result))

    result.extend(zero_list)

    return result

arrange_zeros([1,2,0,0,4,0,5,6])
```

Split a list into chunks of size k

```python
def split_list(arr,k):
    new_list = []
    for x in range(0,len(arr),k):
        new_list.append(arr[x:x+k])
    return new_list

split_list([1, 2, 3, 4, 5, 6, 7], 3)
```
