Merge two dictionaries; if keys overlap, sum their values

```python
def overlap_dicts(dict_1, dict_2):

    new_dict = {}

    for each_key in dict_1:
        if each_key in dict_2.keys():
            new_dict[each_key] = dict_1[each_key] + dict_2[each_key]
        else:
            new_dict[each_key] =  dict_1[each_key]

    for each_key in dict_2:
        if each_key not in new_dict.keys():
            new_dict[each_key] = dict_2[each_key]

    
    return new_dict

overlap_dicts({'a':1,'b':1},{'a':1,'c':3})
```

```python
# This method is much better because it uses only one loop

def overlap_dicts(dict_1, dict_2):
    results = dict_1.copy()

    for x,y in dict_2.items():
        if x in results:
            results[x] = results[x] + y
        else:
            results[x] = y

    return results

overlap_dicts({'a':1,'b':1},{'a':1,'c':3})
```

```python
from collections import Counter

def overlap_dicts(dict_1, dict_2):
    counter_1 = Counter(dict_1)
    counter_2 = Counter(dict_2)

    return dict(counter_1 + counter_2)


overlap_dicts({'a':1,'b':1},{'a':1,'c':3})
```

Invert a dictionary (keys become values, values become keys)

```python
def invert_dictionary(dict_1):
    result_dict = {}

    for x,y in dict_1.items():
        if y not in result_dict:
            result_dict[y] = [x]
        else:
            result_dict[y].append(x)

    return result_dict

invert_dictionary({'a':1,'b':1})
```

Find the key with the maximum value in a dictionary

```python
def find_maximum(dict_1):
    
    return max(dict_1, key=dict_1.get)

find_maximum({'a':1,'b':10})
```

```python
def find_maximum(dict_1):
    max_value = float('-inf')
    max_key = None

    for key,value in dict_1.items():
        if value>max_value:
            max_value = value
            max_key = key

    return max_key

find_maximum({'a':-1,'b':0, 'c':-10})
```

Count word frequency in a sentence and return as a dictionary

```python
def count_characters(sentence):
    words = sentence.lower().split()
    result = {}

    for each_word in words:
        result[each_word] = result.get(each_word,0) + 1

    return result

count_characters("My name is Sakuni")
```

Sort a dictionary by its values in descending order

```python
def sort_dictionary(dict_1):
    sorted_dict = dict(sorted(dict_1.items(), key = lambda x: x[1], reverse=True))
    return sorted_dict

sort_dictionary({'a':-1,'b':0, 'c':-10})
```

```python
def sort_dictionary(dict_1):
    sorted_dict = sorted(dict_1, key = dict_1.get, reverse=True)
    return sorted_dict

sort_dictionary({'a':-1,'b':0, 'c':-10})
```

Find elements that are in list A but not in list B

```python
def set_difference(set_1, set_2):

    return set_1.difference(set_2)

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
set_difference(set1, set2)
```

Check if one set is a subset of another

```python
def check_subset(set_1, set_2):

    return set_1.issubset(set_2)

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
check_subset(set1, set2)
```

 Find all unique pairs from a list that sum to a target value

```python
def find_pairs(arr,val):
    results = []
    seen = set((arr.copy()))

    for each in arr:
        compliment = val - each
        if compliment in seen:
            results.append(tuple((each, compliment)))
            seen.remove(each)
    return results

list = [1,2,3,4,4,5,6,7,8]
target = 8
find_pairs(list, target)
```
