Parse a list of email addresses and extract only the domain names

```python
import re

def find_domain_names1(email_list):

    results = []

    for each in email_list:
        x = re.split("@", each)
        results.append(x[-1])

    return results


```

```python
def find_domain_names2(email_list):

    results = []

    for each in email_list:
        x = each.split("@")
        results.append(x[-1])

    return results
```

```python
email_lst = ['user@gmail.com','admin@yahoo.co.uk','test@company.org']
find_domain_names1(email_lst)
```

```python
find_domain_names2(email_lst)
```

Given a list of dictionaries (e.g., student records), sort them by a specific key

```python
def sort_dictionary(list_1):

    return sorted(list_1, key= lambda x: x["age"])

users = [{'name': 'Bob', 'age': 30}, {'name': 'Alice', 'age': 25},  {'name': 'John', 'age': 20}]

sort_dictionary(users)
```
