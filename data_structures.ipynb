**Two Sum** - Given an array and target, find two numbers that sum to target

```python
def find_two_numbers(arr, target):

    for each in arr:
        other = target - each
        if other in arr:
            return tuple((each, other))

    return None

find_two_numbers([1,4,5],6)
```

```python
# Issues in above function
# - Same element can be used twice due to if other in arr: loop
# - if other in arr: is an ineffecient lookup 
# if other in arr is slow because Python must scan the list element by element, 
# if complement in seen is fast because Python uses hashing for constant-time lookup.
```

```python
def find_two_numbers(arr, target):
    seen = set((arr.copy()))

    for each in arr:
        compliment = target - each
        seen.remove(each)
        if compliment in seen:
            return (each, compliment)

    return None

find_two_numbers([1, 2, 4, 5], 6)
```

```python
# Issues in above function
# - Creating a set in the beginning loses duplicates. This leads to incorrect results (ex: [3,3],6)
# - remove() can lead to KeyErrors when same value appears more times in list
# - More complex than needed due to preloading the set and mutating during iteration
```

```python
def find_two_numbers(arr, target):
    seen = set()

    for num in arr:
        compliment = target - num
        if compliment in seen:
            return (num, compliment)
        seen.add(num)
    return None

find_two_numbers([1, 2, 4, 5], 6)
```

**Valid Anagram** - Check if two strings are anagrams

```python
from collections import Counter

def if_anagram(string1, string2):
    s1 = string1.lower().replace(" ","")
    s2 = string2.lower().replace(" ","")

    if len(s1) != len(s2):
        return False

    if (Counter(s1) == Counter(s2)):
        return True

    return False


if_anagram('spar','rasp')
```

```python
# Above function has
# - early length check
# - Counter creates a frequency map, and a string can only be an anagram if counters are equal
# - Time complexity id O(n), therefore very effecient
```

```python
def if_anagram(string1, string2):
    s1 = string1.lower().replace(" ","")
    s2 = string2.lower().replace(" ","")

    return sorted(s1) == sorted(s2)

if_anagram('spar','rasp')
```

```python
# Above function creates a sorted list of both strings and compare them
# Anagrams should have the same sorted list. Therefore, code is simpler.
# Time complexity is O(n log n), therefore slower
```

```python
def if_anagram(string1, string2):
    s1 = string1.lower().replace(" ","")
    s2 = string2.lower().replace(" ","")

    freq = {}

    if len(s1) != len(s2):
        return False

    for char in s1:
        freq[char] = freq.get(char, 0) + 1

    for char in s2:
        if char not in freq or freq[char] == 0:
            return False
        freq[char] -= 1
    return True

if_anagram('sparr','rasp')
```

Reverse a String - Reverse string in-place with O(1) space

```python
def reverse_string(s1):

    new_string = ""

    for each in s1[::-1]:
        new_string += each
    return new_string

reverse_string("Hello World")
```

```python
# Above answer is incorrect because a new data structure is created instead of modifying the existing one
# Using O(1) space means that extra memory is contant and no new strings, arrays are made

# A string cannot be reversed in place because strings are immutable.
# Only mutable structures can have in-place reversal

# AIM ---> Reverse a list of characters in place
```

```python
def reverse_string(s1):

    new_string = list(s1)

    return "".join(new_string[::-1])
    

reverse_string("Hello World")
```

```python
# Above is also wrong since slicing also create a copy in Python.
# So, the in-place constraint is violated

# join also takes O(n) space. Therefore, that should be avoided as well
```

```python
def reverse_string(chars):
    left = 0
    right = len(chars) - 1

    while left < right:
        chars[left], chars[right] = chars[right], chars[left]
        left += 1
        right -= 1
    
    return chars

s = list("Hello World!")
reverse_string(s)
print("".join(s))
```
