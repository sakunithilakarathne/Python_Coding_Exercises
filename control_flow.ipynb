Write a function to check if a number is prime

```python
def check_prime_num(number):
    count = 0
    for each in range(1, number+1):
        if number % each == 0:
            count += 1
            if count > 2:
                return False
            
    return True

check_prime_num(3)
```

```python
def check_prime_num(number):
    if number <= 1:
        return False
    
    for each in range(2, number):
        if number % each == 0:
            return False
            
    return True

check_prime_num(3)
```

Generate the first n numbers in the Fibonacci sequence

```python
def fibonacci_numbers(val):
    first_num = 0
    second_num = 1
    results = [0,1]

    for each in range(val-1):
        seq = first_num + second_num
        results.append(seq)
        first_num = second_num
        second_num = seq
    return results

fibonacci_numbers(9)

```

```python
def fibonacci_numbers(n):
    if n <= 0:
        return []
    if n == 1:
        return [0]

    results = [0, 1]

    for _ in range(n - 2):
        results.append(results[-1] + results[-2])

    return results

fibonacci_numbers(9)
```

Find the factorial of a number using recursion

```python
def find_factorial(n):
    result = 1

    if n < 0:
        return False

    if n == 0 or n == 1:
        return 1
    
    for each in range(2,n+1):
        result *= each
    
    return result

find_factorial(10)
```

```python
def find_factorial(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers")

    if n == 0 or n == 1:
        return 1

    return n * find_factorial(n - 1)

find_factorial(10)
```

Check if a number is a perfect square without using sqrt

```python
def check_squareroot(number):

    sqr = number ** 0.5 
    sqr1 = sqr % 10

    if sqr % 10 != 0:
        return False
    
    return sqr, sqr1

check_squareroot(10)
    
```
