Write a function that reverses the words in a sentence (not the characters)

```python
def reverse_string(sentence):
    words = sentence.split()
    print(" ".join(words[::-1]))

sentence = "It is a cat"
reverse_string(sentence)
```

```python
def reverse_sentence(sentence):
    word_list = sentence.split()
    word_list.reverse()
    print(" ".join(word_list))

reverse_sentence("Sky is Blue")
```

Count the frequency of each character in a string and return as a dictionary

```python
def count_characters(string):
    freq = {}
    letters = string.replace(" ", "").lower()
    for letter in letters:
        if letter in freq:
            count = freq[letter] + 1
            freq.update({letter:count})
        else:
            freq.update({letter:1})
    return freq

count_characters("I am Isabell")
```

```python
def count_characters(string):
    freq = {}
    characters = string.replace(" ", "").lower()
    for char in characters:
        freq[char] = freq.get(char,0) + 1
    return freq
        


count_characters("I am Isabell")
```

Check if two strings are anagrams of each other

```python
def check_anagram(string1, string2):
    string1_set = list(string1.lower())
    for char in string2.lower():
        if len(string1) == len(string2) and char in string1_set:
            string1_set.remove(char)
        else:
            print(f"{string1} is not an anagram of {string2}")
            break
        print(f"{string1} is an anagram of {string2}")

check_anagram("palm", "lamp")
```

```python
def check_anagram(string_1, string_2):

    string_1 = string_1.replace(" ","").lower()
    string_2 = string_2.replace(" ","").lower()

    if len(string_1) != len(string_2):
        return False
    
    char_counter = {}

    for char in string_1:
        char_counter[char] = char_counter.get(char, 0) + 1
    
    for char in string_2:
        if char not in char_counter:
            return False
        char_counter[char] -= 1
        if char_counter[char] < 0:
            return False
    return True

print(check_anagram("palm", "lamp"))
```

```python
from collections import Counter

def check_anagram(string_1, string_2):
    return Counter(string_1.lower()) == Counter(string_2.lower())

print(check_anagram("palm", "lamp"))
```

 Find the first non-repeating character in a string

```python
def find_non_repeating_character(string):

    char_set = {}

    string = string.replace(" ","").lower()
    for char in string:
        char_set[char] = char_set.get(char, 0) + 1
    
    for each in char_set:
        if char_set[each] == 1:
            print(f"{each} is the first non repeating character")
            break

find_non_repeating_character("swiss")
```
